```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

accel_linear = pd.read_csv('data/iPhone Thing-Accelerometer_Linear.csv')
accel_x = pd.read_csv('data/iPhone Thing-Accelerometer_X.csv')
accel_x = accel_x.rename(columns={'value':'accelerometer_x'})

accel_y = pd.read_csv('data/iPhone Thing-Accelerometer_Y.csv')
accel_y = accel_y.rename(columns={'value':'accelerometer_y'})

accel_z = pd.read_csv('data/iPhone Thing-Accelerometer_Z.csv')
accel_z = accel_z.rename(columns={'value':'accelerometer_z'})

gps = pd.read_csv('data/iPhone Thing-Gps.csv')

accel_x["time"] = pd.to_datetime(accel_x["time"])
accel_y["time"] = pd.to_datetime(accel_y["time"])
accel_z["time"] = pd.to_datetime(accel_z["time"])


data_full = [accel_linear,accel_x,accel_y,accel_z,gps]

```

### Accelerometer_X vs Time

```python
sns.scatterplot(data=accel_x,x='time',y='accelerometer_x',legend='auto')
```

### Acceleromter_Y vs Time

```python
sns.scatterplot(data=accel_y,x='time',y='accelerometer_y',legend='auto')
```

### Accelerometer_Z vs Time

```python
sns.scatterplot(data=accel_z,x='time',y='accelerometer_z',legend='auto')
```

### Accelerometer X,Y, Z vs Time

```python
plt.figure(figsize=(12, 6))

sns.scatterplot(data=accel_x,x='time',y='accelerometer_x',label='x')
sns.scatterplot(data=accel_y,x='time',y='accelerometer_y',label='y')
sns.scatterplot(data=accel_z,x='time',y='accelerometer_z',label='z')

plt.title("Accelerometer X, Y and Z")
plt.xlabel("Time")
plt.ylabel("Acceleration")
plt.xticks(rotation=45)
plt.legend(title="Axis")

plt.tight_layout()
plt.show()
```

### Accelerometer X, Y,Z 3D

```python


min_len = min(len(accel_x['accelerometer_x']),len(accel_y['accelerometer_y']), len(accel_z['accelerometer_z']))

fig = plt.figure(figsize=(9, 7))
ax = fig.add_subplot(projection='3d')

scatter = ax.scatter3D(accel_x['accelerometer_x'][:min_len], accel_y['accelerometer_y'][:min_len], accel_z['accelerometer_z'][:min_len])

# Labels for all three dimensions
ax.set_xlabel('Accelerometer X')
ax.set_ylabel('Accelerometer Y')
ax.set_zlabel('Accelerometer Z')
ax.set_title('Accelerometer X,Y,Z')

plt.tight_layout()
plt.show()
```
